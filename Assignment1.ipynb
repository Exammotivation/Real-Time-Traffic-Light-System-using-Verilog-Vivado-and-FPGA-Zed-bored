```python
#Q2. Take two numbers as input from the user
a = int(input("a: "))
b = int(input("b: "))

# 2. Create an empty list to store the prime numbers we find
primes = []

# 3. Check every number in the range from a to b (inclusive)
for num in range(a, b + 1):
    if num > 1:  # Prime numbers must be greater than 1
        is_prime = True  # Assume the number is prime first
        
        # Check if num can be divided by any number from 2 up to num-1
        for i in range(2, num):
            if num % i == 0:
                is_prime = False  # Found a divisor, so it is not prime
                break  # Stop checking this number
                
        if is_prime:
            primes.append(num)  # Add the prime number to our list

# 4. Print the results
print(f"\nPrime numbers in the range [{a}, {b}]:", primes)
print("Total prime numbers found:", len(primes))
```

```python
#Q3
import numpy as np

# 1. Generate 12 random integers between 20 and 80 using NumPy
original_array = np.random.randint(20, 81, size=12)

# 2. Create an index array from 0 to 11
original_indices = np.arange(12)

# 3. Combine them into a 2D array: row 0 has numbers, row 1 has original indices
paired_matrix = np.vstack((original_array, original_indices))

# 4. Bubble Sort the NumPy matrix in descending order based on row 0 (the numbers)
n = paired_matrix.shape[1]  # Get the number of columns (12)
for i in range(n):
    for j in range(0, n - i - 1):
        # Compare numbers in the top row (row 0)
        if paired_matrix[0, j] < paired_matrix[0, j + 1]:
            # Swap the entire columns (both the number and its index swap together)
            paired_matrix[:, [j, j + 1]] = paired_matrix[:, [j + 1, j]]

# 5. Extract the sorted results back out
sorted_list = paired_matrix[0, :].tolist()
sorted_indices = paired_matrix[1, :].tolist()

# 6. Print the results
print("Original List:    ", original_array.tolist())
print("Sorted List:      ", sorted_list)
print("Original Indices: ", sorted_indices)
```

```python
#Q7. (Probability Simulation– Coin & Dice)
#(a) Simulate a fair coin toss 5000 times. Calculate the experimental probability of Heads and Tails.
#(b) Simulate rolling a fair die 8000 times. Print the count and probability of each face (1to 6).
#(c) Write a short comment comparing experimental probability with theoretical probability

import numpy as np


# (a) Coin Simulation (5000 flips)

# 0 means Tails, 1 means Heads
coin = np.random.randint(0, 2, size=5000)

heads = np.sum(coin == 1)
tails = np.sum(coin == 0)
print("Coin toss result")
print(heads)
print(" Probability:", heads / 5000)
print(tails)
print(" Probability:", tails / 5000)

# (b) Dice Simulation (8000 rolls)

# Generates 8000 random rolls from 1 to 6
dice = np.random.randint(1, 7, size=8000)

# Count how many times each face appears
face1 = np.sum(dice == 1)
face2 = np.sum(dice == 2)
face3 = np.sum(dice == 3)
face4 = np.sum(dice == 4)
face5 = np.sum(dice == 5)
face6 = np.sum(dice == 6)

print("Dice flip result")
print(face1)
print(" Probability:", face1 / 8000)
print(face2)
print(" Probability:", face2 / 8000)
print(face3)
print(" Probability:", face3 / 8000)
print(face4)
print(" Probability:", face4 / 8000)
print(face5)
print(" Probability:", face5 / 8000)
print(face6)
print(" Probability:", face6 / 8000)
# ==========================================
# (c) Simple Comparison Explanation
# ==========================================
# Theoretical probability is what should happen in perfect math (like 50% for coins).
# Experimental probability is what actually happens in our real test run.
# Because we ran the test thousands of times, our experiment matches the math perfectly.
```

```python
#Q9

# Step 1: Create the initial CSV file with 60 students

# Generate 60 rows and 3 columns of random integer marks between 30 and 100
# np.random.randint(low, high, size) includes 30 up to 100 (high is 101)

marks_data = np.random.randint(30, 101, size=(60, 3))

# Define column names
subjects = ['Maths', 'Physics', 'Chemistry']

# Create a DataFrame
df_students = pd.DataFrame(marks_data, columns=subjects)

# Save the generated data to 'student_marks.csv'
df_students.to_csv('student_marks.csv', index=False)
print("Initial 'student_marks.csv' file created successfully!\n")

# Step 2: Read the file using Pandas

# Load the data from the CSV file into a Pandas DataFrame
df = pd.read_csv('student_marks.csv')

# Step 3: Compute and print average marks for each subject

# The .mean() function calculates the arithmetic mean for each numerical column
average_marks = df[['Maths', 'Physics', 'Chemistry']].mean()

print("--- Average Marks for Each Subject ---")
print(average_marks)
print("-" * 40)

# Step 4: Find how many students scored > 75 in Maths


# df['Maths'] > 75 creates a series of True/False values.
# .sum() counts how many 'True' values exist.
maths_above_75 = (df['Maths'] > 75).sum()

print(f"Number of students who scored > 75 in Maths: {maths_above_75}")
print("-" * 40)

# Step 5: Add a 'Total' column and save updated data


# Add the three subject columns together row-by-row
df['Total'] = df['Maths'] + df['Physics'] + df['Chemistry']

# Display the first 6 rows of the updated dataset to verify
print("Preview of Updated Data (First 6 Rows):")
print(df.head(6))
print("-" * 40)

# Save the updated DataFrame into a new CSV file
df.to_csv('updated_student_marks.csv', index=False)
print("Updated dataset saved to 'updated_student_marks.csv'!")
```

```python
#Q10

# (a) Plot y = 2x^2 - 5x + 3 for x in [-8, 8]


# 1. Generate 400 evenly spaced x-values between -8 and 8
x = np.linspace(-8, 8, 600)

# 2. Compute y-values using vector arithmetic
y = 2 * (x ** 2) - 5 * x + 3

# 3. Create figure for Part (a)
plt.figure(figsize=(8, 5))

# Plot x vs y with a blue line
plt.plot(x, y, label=r'$y = 2x^2 - 5x + 3$', color='blue', linewidth=1)

# Add title, axis labels, grid, and legend
plt.title('Plot of Quadratic Function: $y = 2x^2 - 5x + 3$', fontsize=14)
plt.xlabel('x-axis', fontsize=12)
plt.ylabel('y-axis', fontsize=12)
plt.grid(True, linestyle='--', alpha=0.5)
plt.legend(fontsize=11)

# Display the plot
plt.show()


# (b) Histogram of 500 Normal Random Numbers


# 1. Generate 500 numbers from normal distribution (mean = 5, std dev = 2)
mean_val = 5
std_dev = 2
data = np.random.normal(loc=mean_val, scale=std_dev, size=500)

# 2. Create figure for Part (b)
plt.figure(figsize=(8, 5))

# 3. Plot histogram with 25 bins
plt.hist(data, bins=25, edgecolor='black', color='skyblue', alpha=0.7)

# 4. Calculate actual sample mean and draw a vertical dashed line
sample_mean = np.mean(data)
plt.axvline(sample_mean, color='red', linestyle='--', linewidth=2, 
            label=f'Mean = {sample_mean:.2f}')

# Add title, axis labels, grid, and legend
plt.title('Histogram of Normal Distribution (Mean = 5, Std Dev = 2)', fontsize=14)
plt.xlabel('Value', fontsize=12)
plt.ylabel('Frequency (Count)', fontsize=12)
plt.grid(True, linestyle='--', alpha=0.5)
plt.legend(fontsize=11)

# Display the plot
plt.show()
```
